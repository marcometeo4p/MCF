## 5 - <ins>Dinamica Sistemi Planetari</ins>

Dato un sistema ad N corpi soggetto a forze gravitazionali possiamo combinare la seconda legge della dinamica con la legge di gravitazione universale di Newton per calcolare la forza agente sul corpo $i$ e la sua conseguente accelerazione:

$$
m_i \mathbf{a_i} = - \sum_{j \neq i}^j G \frac{m_i m_j}{|\mathbf{r_{ij}}|^2} \frac{\mathbf{r_{ij}}}{|\mathbf{r_{ij}}|}
$$

Semplifcando $m_i$ ed esplicitando una singola componente ($x$ in particolare) abbiamo:

$$
a_{i_x} = \sum_{j \neq i}^j G \frac{ m_j}{  \left( (x_i - x_j)^2 + (y_i - y_j)^2 +(z_i - z_j)^2  \right)^{3/2}} (x_j - x_i) \hat{x} 
$$

Partendo da condizioni iniziali note (posizioni e velocità), si può calcolare l'evoluzione del sistema procedendo per passi.
