```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dati=pd.read_csv("ExoplanetsPars_2025.csv",comment="#")
print(dati.columns)
dati.head(10)
```

```python
'''
pl_orbper: Orbital Period [days]
pl_bmassj: Planet Mass or Mass*sin(i) [Jupiter Mass]
pl_orbsmax: Orbit Semi-Major Axis [au]
st_mass: Stellar Mass [Solar mass]
discoverymethod: Discovery Method
'''
plt.figure(figsize=(12,8))
plt.scatter(dati["pl_orbper"], dati["pl_bmassj"],s=15,c="cornflowerblue" ,label="dati")
plt.xlabel('Orbital Period [days]')
plt.xscale('log')
plt.yscale('log')
plt.legend()
plt.title("Flusso di elettroni",fontsize=15)
plt.ylabel('Planet mass [Jupiter Mass]')
#plt.savefig("graf_parz.png")
#plt.savefig("graf_parz.pdf")
```

```python
plt.figure(figsize=(12,8))
plt.scatter(dati['pl_orbper'],(dati["pl_orbsmax"]**2)/ dati["st_mass"],s=15,c="cornflowerblue" ,label="exoplanets")
plt.xlabel('Orbital Period [days]')
plt.xscale('log')
plt.yscale('log')
plt.legend()
#plt.title("Flusso di elettroni",fontsize=15)
plt.ylabel('$R_{max}^2/m_{*} [m^2/days]$')
#plt.savefig("graf_parz.png")
#plt.savefig("graf_parz.pdf")
```
